```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

# Sample Data
data = {
    'Age': [22, 25, 47, 52, 46, 56, 36, 32, 26, 27, 45, 41, 23, 24, 28, 30, 35, 55, 60, 65],
    'Gender': ['Male', 'Female', 'Female', 'Male', 'Female', 'Male', 'Female', 'Female', 'Male', 'Male',
               'Female', 'Male', 'Male', 'Female', 'Female', 'Male', 'Male', 'Female', 'Male', 'Female']
}
df = pd.DataFrame(data)
print(df)

# Improved Age Histogram
plt.figure(figsize=(10,6))
sns.histplot(df['Age'], bins=[20, 30, 40, 50, 60, 70], kde=True, color='lightcoral', edgecolor='black')
plt.title('Age Group Distribution', fontsize=16)
plt.xlabel('Age Range', fontsize=14)
plt.ylabel('Number of People', fontsize=14)
plt.xticks([20, 30, 40, 50, 60, 70])
plt.grid(axis='y')
plt.show()
```

```python
# Step 1: Install and Import
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

# Step 2: Dataset
data = {
    'Country': ['India', 'China', 'United States', 'Indonesia', 'Pakistan', 'Brazil', 'Nigeria', 'Bangladesh', 'Russia', 'Mexico'],
    'Population': [1393409038, 1444216107, 331893745, 273523621, 220892331, 213993437, 206139587, 166303498, 145912025, 130262216]
}
df = pd.DataFrame(data)
print(df)

# Step 3: Plotting - Bar Chart for Population by Country
plt.figure(figsize=(12,6))
sns.barplot(x='Country', y='Population', data=df, hue='Country', palette='viridis', legend=False)
plt.title('Population by Country', fontsize=16)
plt.xticks(rotation=45)
plt.ylabel('Population', fontsize=12)
plt.xlabel('Country', fontsize=12)
plt.grid(True)
plt.tight_layout()  # To ensure the labels fit in the plot area
plt.show()

# Step 4: Plotting - Histogram for Distribution of Country Populations
plt.figure(figsize=(8,6))
sns.histplot(df['Population'], bins=5, kde=True, color='teal')
plt.title('Distribution of Country Populations', fontsize=16)
plt.xlabel('Population', fontsize=12)
plt.ylabel('Number of Countries', fontsize=12)
plt.grid(True)
plt.tight_layout()  # Ensures the plot components don't overlap
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

# Sample Data for Gender
data = {
    'Gender': ['Male', 'Female', 'Female', 'Male', 'Female', 'Male', 'Female', 'Female', 'Male', 'Male',
               'Female', 'Male', 'Male', 'Female', 'Female', 'Male', 'Male', 'Female', 'Male', 'Female']
}

# Creating DataFrame
df = pd.DataFrame(data)
print(df)

# Gender Distribution - Bar Chart (Fix for FutureWarning)
plt.figure(figsize=(8,6))
sns.countplot(x='Gender', data=df, hue='Gender', palette='viridis', legend=False)
plt.title('Gender Distribution', fontsize=16)
plt.xlabel('Gender', fontsize=14)
plt.ylabel('Number of People', fontsize=14)
plt.grid(True, axis='y')
plt.tight_layout()  # Ensures everything fits properly
plt.show()
```
